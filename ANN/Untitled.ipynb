# ANN 

```python
X = X.astype(int)
```

```python
y = y.replace({"Yes":1,"No":0})
y = y.astype(int)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
Rakhi = Sequential()
```

```python
Rakhi.add(Dense(units=6 , activation="relu",kernel_initializer="uniform"))
Rakhi.add(Dense(units=6 , activation="relu",kernel_initializer="uniform"))
Rakhi.add(Dense(units=6 , activation="relu",kernel_initializer="uniform"))
```

```python
Rakhi.add(Dense(units=1 , activation="sigmoid",kernel_initializer="uniform"))
```

```python
Rakhi.compile(loss = 'mean_squared_error',optimizer = 'sgd', metrics = ["accuracy"])
```

```python
history = Rakhi.fit(X_train, y_train, batch_size=10, epochs=50)
```

```python
loss, accuracy = Rakhi.evaluate(X_test,y_test)
```

```python
y_pred = Rakhi.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```
