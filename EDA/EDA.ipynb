# Feature Scalling
- Feature scalling is a technique to Standarize the independent feautre present in our data to fixed Range.

- Feature Scalling is 2 type

    - Standarization
    - Normalization

# 

### Normalization 

- Normalization in Machine Learning is a preprocessing technique that rescales numerical features to a standard range (usually between 0 and 1). It ensures no single feature dominates the model’s learning simply because it has larger values, ultimately improving algorithm accuracy and training speed.

### Standarization
- Standardization in machine learning is a preprocessing technique that rescales numerical features so they have a mean of zero and a standard deviation of one. It ensures features contribute equally to model training, preventing variables with larger scales from disproportionately biasing the algorithm
